```python
import numpy as np
from config import config
from simulator import Simulator
from replacement_agent import ReplacementAgent
from order_agent import OrderAgent
import my_utils as mu
```

#### Create data and store it

```python
# Create simulator object 
S = Simulator(cf=config)
```

```python
# create failure samples and corresponding mean ln RUL parameters
Ns = 50
tfs = S.draw_tf_samples(Ns=Ns)
muks = S.mu_ln_RUL_pred_k_given_tf_vec(tf_vec=tfs)
```

#### Evaluate a certain heuristic on the data

```python
# load data
tfs = mu.load_file(filename="./data/tf_samples", file_type="h5")
muks = mu.load_file(filename="./data/mus", file_type="h5")
```

```python
# create ReplacementAgent object
A = ReplacementAgent(cf=config)
# define heuristic policy
fun = "rh1"
```

```python
# evaluate heuristic for setup defined in the config
costs, tlcs, R, var_R = A.eval_rh(fun=fun, tfs=tfs, muks=muks)
print(f"heuristic performance: {R}")
print(f"Variance of performance: {var_R}")
```

#### Evaluate a certain heuristic on a range of cost ratios

```python
# choose cost ratios for evaluation
cost_ratios = np.array([0.01, 0.1])
# evaluate heuristic for all chosen cost ratios
R, var_R, _, _ = A.eval_mult_c_ratios(c_rs=cost_ratios, fun=fun, tfs=tfs, mu_ln_RUL_preds=muks, calc_M=False)

print(f"heuristic performances: {R}")
print(f"performance variances: {var_R}")
```
